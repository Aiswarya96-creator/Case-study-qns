```python
class bankaccount:
  def __init__(self,owner,acc_no,balance):
    self.owner=owner
    self.acc_no=acc_no
    self.balance=balance

  def deposit(self,amount):
      self.balance+=amount
      print(f"Deposited:{amount:.2f}, Balance:{self.balance:.2f}")

  def withdraw(self,amount):
    if self.balance>=amount:
      self.balance-=amount
      print(f"withdrew:{amount:.2f},Balance:{self.balance:.2f}")
    else:
      print("insufficient balance")

  def display(self):
    print(f"owner:{self.owner},acc_no:{self.acc_no},balance:{self.balance}")

cust1=bankaccount("Aishu",7890,50000)
cust2=bankaccount("daya",6789,0)
cust1.display()
cust2.display()
cust2.withdraw(1000)
cust1.withdraw(20000)

```
